## 1. Variables & Data Types

**Key Difference from Java/C++:** No type declarations needed. Python figures out the type automatically.

| Type | Example | Notes |
|---|---|---|
| Integer | `age = 25` | Same as `int` |
| Float | `height = 5.9` | Same as `double` |
| String | `name = "Alice"` | Text, use quotes |
| Boolean | `is_student = True` | `True` or `False` (capital T/F!) |
| None | `value = None` | Like `null` in Java |

````python
# Example: Variables (no type declaration needed)
age = 25
height = 5.9
name = "Alice"
is_student = True

print(age)
print(height)
print(name)
print(is_student)

# Check the type
print(type(age))  # <class 'int'>
print(type(height))  # <class 'float'>
print(type(name))  # <class 'str'>
````

## 2. Strings & String Methods

Strings in Python are very powerful. You can call methods directly on them.

| Method | What it does | Example |
|---|---|---|
| `.upper()` | Convert to UPPERCASE | `"hello".upper()` → `"HELLO"` |
| `.lower()` | Convert to lowercase | `"HELLO".lower()` → `"hello"` |
| `.strip()` | Remove spaces from ends | `"  hello  ".strip()` → `"hello"` |
| `.split()` | Split into list | `"a,b,c".split(",")` → `["a", "b", "c"]` |
| `.replace()` | Replace text | `"hello".replace("h", "j")` → `"jello"` |
| `len()` | Length of string | `len("hello")` → `5` |

````python
# Example: String methods
text = "  Hello World  "

# Remove spaces from ends
cleaned = text.strip()
print(f"Original: '{text}'")
print(f"Cleaned: '{cleaned}'")

# Convert to uppercase
upper = cleaned.upper()
print(f"Uppercase: {upper}")

# Get length
length = len(cleaned)
print(f"Length: {length}")

# Split by space
words = cleaned.split(" ")
print(f"Words: {words}")
````

## 3. Lists (Like Arrays)

Lists are ordered collections. They're like arrays in Java, but more flexible.

| Operation | Example | Result |
|---|---|---|
| Create list | `my_list = [1, 2, 3]` | `[1, 2, 3]` |
| Access element | `my_list[0]` | `1` (0-indexed, like Java) |
| Add element | `my_list.append(4)` | `[1, 2, 3, 4]` |
| Length | `len(my_list)` | `4` |
| Slice | `my_list[0:2]` | `[1, 2]` (up to index 2, not including 2) |

````python
# Example: Lists
fruits = ["apple", "banana", "cherry"]

# Access by index
print(fruits[0])  # "apple"

# Add element
fruits.append("date")
print(fruits)  # ["apple", "banana", "cherry", "date"]

# Get length
print(len(fruits))  # 4

# Slice (first 2 elements)
first_two = fruits[0:2]
print(first_two)  # ["apple", "banana"]
````

## 4. Loops (for loop)

**Key Difference from Java:** Python's `for` loop works like Java's enhanced for loop by default.

```python
# Python
for item in my_list:
    print(item)

# This is equivalent to Java's:
// Java
for (String item : myList) {
    System.out.println(item);
}
```

````python
# Example: For loop over a list
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)

print("\n--- Using range ---")
# If you need index, use range()
for i in range(3):  # 0, 1, 2
    print(f"Index {i}: {fruits[i]}")
````

## 5. Conditionals (if/else)

**Key Difference:** No parentheses needed, and no curly braces. Use **indentation** instead!

````python
# Example: if/else
age = 25

if age < 18:
    print("You are a minor")
elif age >= 18 and age < 65:
    print("You are an adult")
else:
    print("You are a senior")

# Also works with strings
name = "Alice"
if name == "Alice":
    print("Hello Alice!")
````

## 6. Functions

Functions are defined with `def` keyword. They are similar to Java methods but simpler.

````python
# Example: Simple function
def greet(name):
    """This function greets someone. (Documentation string)"""
    message = f"Hello, {name}!"
    return message

result = greet("Bob")
print(result)

# Function with multiple parameters
def add_numbers(a, b):
    return a + b

sum_result = add_numbers(5, 3)
print(f"5 + 3 = {sum_result}")

# Function that takes a list
def print_list(items):
    for item in items:
        print(f"  - {item}")

print_list(["apple", "banana", "cherry"])
````

## 7. Type Conversion

Converting between data types is very common in the capstone.

````python
# Example: Type conversion
price_string = "15.50"
price_float = float(price_string)  # Convert string to float
print(f"Type: {type(price_float)}, Value: {price_float}")

# String to int
count_string = "42"
count_int = int(count_string)
print(f"Type: {type(count_int)}, Value: {count_int}")

# Number to string
age = 25
age_string = str(age)
print(f"Type: {type(age_string)}, Value: {age_string}")

# Combining: Remove '$' and convert to float
messy_price = "$19.99"
clean_price = float(messy_price.replace("$", ""))
print(f"Clean price: {clean_price}")
````

## 8. Dictionaries (Like HashMaps)

Dictionaries store key-value pairs. Similar to Java's `HashMap<String, Object>`.

````python
# Example: Dictionaries
person = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}

# Access by key
print(person["name"])  # "Alice"

# Add new key-value
person["job"] = "Engineer"
print(person)

# Loop through dictionary
for key in person:
    print(f"{key}: {person[key]}")
````

## 9. Sets (Unique Values Only)

Sets automatically remove duplicates.

````python
# Example: Sets
numbers = [1, 2, 2, 3, 3, 3, 4]

# Convert to set to remove duplicates
unique_numbers = set(numbers)
print(unique_numbers)  # {1, 2, 3, 4}

# Check if item is in set (very fast!)
if 2 in unique_numbers:
    print("2 is in the set")

# Add to set
unique_numbers.add(5)
print(unique_numbers)  # {1, 2, 3, 4, 5}
````
